```python
import  torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from torch.optim import AdamW
```

```python
MODEL_NAME = "cointegrated/rubert-tiny2"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(
  MODEL_NAME,
  num_labels=3
).to(device)

new_tokens = ["АСКУЭ", "КТП", "РЗА", "СИП", "EnergyTool"]
num_added_toks = tokenizer.add_tokens(new_tokens)

model.resize_token_embeddings(len(tokenizer))

for name, param in model.named_parameters():
  if "layer.0" in name:
    model.requires_grad=False

print("Статус градиентов у слоев")
for name, param in model.named_parameters():
  if "embed" in name or "layer.0" in name or "layer.2" in name:
    print(f"{name:<50} requires_grad={param.requires_grad}")

trainable_params = [p for p in model.parameters() if p.requires_grad]

optimizer = AdamW(trainable_params, lr=2e-5)
print("\nМодель готова к дообучению")
```
