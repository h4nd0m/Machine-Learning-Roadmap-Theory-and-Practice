```python
import numpy as np

np.random.seed(42)

# Генерация 100 квартир
X_raw = np.random.uniform(30, 120, size=100) # Площадь

# Настоящие веса: w_1 = 150 (цена за кв.м), b = 2000 (базовая цена)
true_w = 150
true_b = 2000
noise = np.random.normal(0, 1500, size=100)

y = true_w * X_raw + true_b + noise

# 1. Приводим X к 2D-матрице (100, 1)
X_column = X_raw.reshape(-1, 1)

# 2. Добавляем столбец из единиц СЛЕВА для байаса b (получаем матрицу 100x2)
ones = np.ones((100, 1))
X_matrix = np.hstack([ones, X_column])

print("Форма матрицы X_matrix:", X_matrix.shape) # Должно быть (100, 2)
print("Первые 3 строки матрицы X_matrix:\n", X_matrix[:3])
```

```python
# 1
w_hat = np.linalg.inv(X_matrix.T @ X_matrix) @ X_matrix.T @ y
```

```python
# 2
y_pred = X_matrix @ w_hat
mse = ((y - y_pred)**2).mean()
```

```python
# 3
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_matrix, y)

print(w_hat)
print(model.intercept_, model.coef_[1])
```
