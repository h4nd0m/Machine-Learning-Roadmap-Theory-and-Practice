### Урок 9.1 — Векторизация текста (Часть 1)
#### Токенизация и Словарь (Vocabulary)
##### Пословесная (Word-level): Текст делится по пробелам и пунктуации.
Пример: "Не работает карта!" -> ["не", "работает", "карта", "!"]
##### Посимвольная (Character-level): Текст делится на отдельные буквы и знаки.
##### Подсловесная (Subword-level - BPE, WordPiece): Компромисс, используемый в современности (BERT, GPT). Слова делятся на частые префиксы, корни и суффиксы.
Пример: "перезаблокировать" -> ["пере", "##заблок", "##ировать"]

#### Словарь (Vocabulary)
проблемы каждого метода займут много места, поэтому просто прогугли по названиям
##### 1.One-Hot Encoding (Одногорячее кодирование)
##### 2.Bag-of-Words (BoW / Мешок слов)
##### 3.N-grams (N-граммы)
##### 4.TF-IDF (Term Frequency – Inverse Document Frequency)



```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "карта заблокирована",
    "заблокирована сумма перевода",
    "перевода нет перевода"
]

# Создаем векторную матрицу TF-IDF (Unigrams+Bigrams)
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2), # Использовать отдельные слова и биграммы
    max_features=5000,  # Ограничить размер словаря топ-5000 важных признаков
    sublinear_tf=True   # Сглаживать частоту слов через 1 + log(tf)
)

X = vectorizer.fit_transform(corpus)

print("Размерность полученной матрицы:", X.shape) # [num_docs, num_features]
print("Признаки (словарь):", vectorizer.get_feature_names_out())
```
