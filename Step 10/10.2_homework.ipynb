```python
import torch
import torch.nn as nn
```

```python

# ==========================================
# ЧАСТЬ 1: Модель на базе LSTM
# ==========================================
class LSTMClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_size, num_classes, num_layers=2):
        super().__init__()
        # TODO 1: Инициализируй слой эмбеддингов (не забудь про padding_idx)
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)

        # TODO 2: Инициализируй слой nn.LSTM. Укажи нужные размерности, количество слоев и правило для батча.
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True, num_layers=num_layers)

        # TODO 3: Инициализируй линейный слой для классификации
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        # TODO 4: Пропусти входной тензор через слой эмбеддингов
        emb = self.embedding(x)

        # TODO 5: Пропусти эмбеддинги через слой LSTM. 
        # Напоминание: LSTM возвращает два объекта (один из них — кортеж состояний). Распакуй их правильно.
        out, (h_n, c_n) = self.lstm(emb)

        # TODO 6: Извлеки краткосрочную память только самого верхнего слоя (используй индексацию).
        final_memory = h_n[-1]

        # TODO 7: Пропусти извлеченную память через линейный слой и верни логиты.
        logits = self.fc(final_memory)
        return logits
```

```python
# ==========================================
# ЧАСТЬ 2: Модель на базе GRU
# ==========================================
class GRUClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_size, num_classes, num_layers=2):
        super().__init__()
        # TODO 8: Инициализируй слой эмбеддингов
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        
        # TODO 9: Инициализируй слой nn.GRU (параметры полностью аналогичны LSTM)
        self.gru = nn.GRU(embed_dim, hidden_size, num_layers=num_layers, batch_first=True)

        # TODO 10: Инициализируй линейный слой
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        # TODO 11: Пропусти входной тензор через эмбеддинги
        emb = self.embedding(x)

        # TODO 12: Пропусти эмбеддинги через GRU. 
        # Обрати внимание: подумай, сколько объектов возвращает GRU, и распакуй их.
        out, h_n = self.gru(emb)

        # TODO 13: Извлеки память верхнего слоя (индексация аналогична LSTM).
        final_memory = h_n[-1]

        # TODO 14: Передай память в классификатор и верни логиты.
        logits = self.fc(final_memory)
        return logits
```

```python

# ==========================================
# ЧАСТЬ 3: Проверка моделей
# ==========================================
# TODO 15: Сгенерируй фейковый батч данных (например, 4 предложения по 10 слов) с помощью функции из библиотеки torch.
fake_batch = torch.randint(1, 5000, (4, 10))

# TODO 16: Создай экземпляры обеих моделей (передай им любые логичные параметры, например vocab_size=1000).
model_lstm = LSTMClassifier(vocab_size=5000, embed_dim=100, hidden_size=128, num_classes=4, num_layers=2)
model_gru = GRUClassifier(vocab_size=5000, embed_dim=100, hidden_size=128, num_classes=4, num_layers=2)

# TODO 17: Сделай forward pass батча через обе модели.
logits_lstm = model_lstm(fake_batch)
logits_gru = model_gru(fake_batch)

# TODO 18: Распечатай размерности (shape) результатов от обеих моделей. Убедись, что они одинаковые и правильные.
print(logits_lstm.shape)
print(logits_gru.shape)
```
