```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
# Используем легкую русскоязычную версию GPT
MODEL_NAME = "ai-forever/rugpt3small_based_on_gpt2"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Загружаем модель и токенизатор (это может занять около минуты)...")

# TODO 1: Загрузи токенизатор с помощью метода AutoTokenizer.from_pretrained
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# TODO 2: Загрузи саму модель с помощью AutoModelForCausalLM.from_pretrained.
model = AutoModelForCausalLM(MODEL_NAME).to(device)

# Наша затравка (prompt), которую модель должна продолжить
prompt = "Внедрение нейросетей в банковском секторе приведет к"
print(f"\nЗатравка: {prompt}\n")

# TODO 3: Токенизируй строку prompt.
encoded_input = tokenizer(
    prompt,
    return_tensors="pt"
)
input_ids = encoded_input['input_ids']

print("Генерируем текст...")
with torch.no_grad():
    # TODO 4: Сгенерируй продолжение с помощью метода .generate() у модели.
    generated_ids = model.generate(
        input_ids,
        max_length=50,
        do_sample=True,
        temperature=0.7,
        top_k=50
    )

# TODO 5: Декодируй сгенерированные токены обратно в текст.
result_text = tokenizer.decode(
    generated_ids[0],
    skip_special_tokens=True
)

print("Результат GPT:")
print(result_text)
```
