```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Загрузка данных
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)
```

```python
from xgboost import XGBClassifier

# Инициализация и обучение
model_xgb = XGBClassifier(
    n_estimators=100,      # Количество деревьев
    learning_rate=0.05,    # Скорость обучения
    max_depth=4,           # Глубина (Level-wise)
    random_state=42,
    eval_metric='logloss'  # Метрика для подавления warning
)

model_xgb.fit(X_train, y_train)
y_pred_xgb = model_xgb.predict(X_test)
print("XGBoost Accuracy:", accuracy_score(y_test, y_pred_xgb))
```

```python
from lightgbm import LGBMClassifier

# Инициализация и обучение
model_lgbm = LGBMClassifier(
    n_estimators=100,      # Количество деревьев
    learning_rate=0.05,    # Скорость обучения
    max_depth=4,           # Ограничение глубины
    num_leaves=15,         # Главный параметр! Число листьев (должно быть < 2^max_depth)
    random_state=42,
    verbose=-1             # Отключает отладочные логи в консоли
)

model_lgbm.fit(X_train, y_train)
y_pred_lgbm = model_lgbm.predict(X_test)
print("LightGBM Accuracy:", accuracy_score(y_test, y_pred_lgbm))
```

```python
from catboost import CatBoostClassifier

# Инициализация и обучение
model_cb = CatBoostClassifier(
    iterations=100,        # Аналог n_estimators
    learning_rate=0.05,    # Скорость обучения
    depth=4,               # Аналог max_depth (Oblivious Trees)
    # cat_features=['col1', 'col2'], # Список категориальных колонок (если есть)
    random_seed=42,
    verbose=0              # Отключает лог каждого шага в консоли
)

model_cb.fit(X_train, y_train)
y_pred_cb = model_cb.predict(X_test)
print("CatBoost Accuracy:", accuracy_score(y_test, y_pred_cb))
```
